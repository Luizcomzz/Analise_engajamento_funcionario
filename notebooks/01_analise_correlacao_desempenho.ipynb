```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data/Pesquisa_de_Clima.csv")

# Remove espaços no início e no final dos nomes das colunas
df.columns = df.columns.str.strip()

# Converte para float com segurança (transforma textos inválidos em NaN)
df['Anos_na_Empresa'] = pd.to_numeric(df['Anos_na_Empresa'], errors='coerce')

df.columns.tolist()
```

```python
variaveis_numericas = [
    'Idade',
    'Nivel_Satisfacao',
    'Anos_na_Empresa',
    'Recomendaria'
]

```

```python
matriz_correlacao = df[variaveis_numericas].corr()

matriz_correlacao
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    matriz_correlacao,
    annot=True,
    fmt=".2f"
)

plt.title("Matriz de Correlação - People Analytics")

plt.show()
```
